```python
### 간단한 선형 회귀를 수행해 보자
import numpy as np
from sklearn import linear_model  # scikit-learn 모듈을 가져온다
```

```python
regression = linear_model.LinearRegression() #클래스 생성자
```

```python
X = [[163], [179], [166], [169], [170]] # << 행렬
```

```python
y = [54, 63, 57, 56, 58] #레이블, 타겟표, 정답지
```

```python
regression.fit(X, y)
```

```python
coef = regression.coef_            # 직선의 기울기
intercept = regression.intercept_  # 직선의 절편regression.intercept()
score = regression.score(X, y)


print(f"y = {coef.round(2)}* X + {intercept:.2f}")
print(f"데이터와 선형회귀 직선의 관계점수: {score:.1%}")
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y, color='blue', maker = 'D')
x = np.arange(162, 180)
y1 = 0.54 * x - 33.17
plt.plot(x, y1, color='red')
plt.show()
```

```python
#경사 하강법
X = np.array([1, 4.5, 9, 10, 13]) #입력 데이터
y = np.array([0, 0.2, 2.5, 5.4, 7.3]) #결과값

w, b = 0, 0 #초기값
learning_rate, epoch = 0.005, 1000 #학습률과 학습 횟수(에폭?)
n = len(X) #입력 데이터 개수
for i in range(epoch):
    y_pred = w*X + b
    error = y_pred - y
    w = w - learning_rate * (error * X).sum()
    b = b - learning_rate * error.sum()

print('w=', w.round(2), ', b =', b.round(2))
```

```python
from sklearn import linear_model
import numpy as np

X = np.array([1, 4.5, 9, 10, 13]) #입력 데이터
y = np.array([0, 0.2, 2.5, 5.4, 7.3]) #결과값
regr = linear_model.LinearRegression()
X = X[:, np.newaxis]
regr.fit(X, y)

print(f'w = {regr.coef_.round(2)}, b = {regr.intercept_.round(2)}')
```
