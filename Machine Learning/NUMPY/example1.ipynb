```python
def train_test_split():
    a = np.arange(1, 51)
    np.random.shuffle(a)
    train_Data = a[:40]
    test_Data = a[40:]

    return train_Data, test_Data

(X_train, X_test) = train_test_split()
print(X_train)
print(X_test)
```

```python
a = np.array([[1, 2], [1, -3]]) # 아 선형 방적식 풀어주는 거구나 #x+2y / x - 3y
b = np.array([6, 1]) #각각 값이 x+2y=6 / x-3y=1
s = np.linalg.solve(a, b) #위 처럼 풀어주는거 !!!
print(a)
print(b)
print(s)
```
