```python
x = np.arange(-20, 21)
y1 = 2 * x
y2 = (1/3) * (x ** 2) + 5
y3 = -(x ** 2) - 5
plt.plot(x, y1, 'g.--') #초록색 큰점에 점선
plt.plot(x, y2, 'r^-') #빨간색 삼각형에 선
plt.plot(x, y3, 'b*:') #파란색 별에 작은 점
#plt.plot(x, y1, 'g--', x, y2, 'r^', x, y3, 'b*') <<이런 식으로 작성 해도 됨
plt.axis([-30, 30, -30, 30])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
y1 = np.sin(x)
y2 = np.cos(x)

plt.style.use('seaborn-v0_8-whitegrid')
plt.title("sin cos curve")
plt.plot(x, y1, 'r', label='sin')
plt.plot(x, y2, 'g:', label='cos')
plt.savefig("sin_cos.png") #.jpg 도 가능

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
